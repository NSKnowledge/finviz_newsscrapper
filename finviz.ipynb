```python
from urllib.request import urlopen, Request
from bs4 import BeautifulSoup
```

```python
finviz_url ='https://finviz.com/quote.ashx?t='
tickers =['AAPL']
```

```python
news_tables={}
for ticker in tickers:
    url = finviz_url + ticker
    
    req = Request(url =url, headers ={'user-agent':'my-app'})
    response = urlopen(req)
    html = BeautifulSoup(response, 'html')
    news_table =html.find(id='news-table')
    news_tables[ticker] =news_table
print(news_tables)
```

```python
amzn_data = news_tables['AAPL']
amzn_rows = amzn_data.findAll('tr')
filtered_rows = []
# for index, row in enumerate(amzn_rows):
#     title = row.a.text    
#     timestamp =row.td.text
#     url = row.a['href'] if row.a else 'No URL'
    

for index, row in enumerate(amzn_rows):
    # Check if <a> and <td> tags exist before extracting their values
    title = row.a.text if row.a else 'No Title'
    timestamp = row.td.text if row.td else 'No Timestamp'
    url = row.a['href'] if row.a else 'No URL'

    # Filter for rows with 'finance.yahoo.com' in the URL
    if 'finance.yahoo.com' in url:
        filtered_rows.append((timestamp, title, url))

# Print filtered rows
for timestamp, title, url in filtered_rows:
    print(f"{timestamp} {title} {url}")
    print(timestamp+ " " +title+ " "+url)
```

```python
from typing import Optional
def fetch_url_content(url:str)->Optional[str]:
    prefix_url: str ="https://r.jina.ai/"
    full_url:str = prefix_url + url
        
    try:
        response = requests.get(full_url)
        if response.status_code ==200:
            return response.content.decode('utf-8')
        else:
            print(f"Error:HTTP Getrequest failed with status code {response.status_code}")
            return None
    except requests.RequestException as e:
        print(f"Error: failed to fetch URL {full_url}. Exception: {e}")
        return None
```

```python
import re

def extract_article_content(raw_text: str) -> str:
    # Step 1: Remove metadata like Title, URL, Published Time, and Markdown Content header
    raw_text = re.sub(r"Title:.*?\n", "", raw_text)  # Remove "Title: ..."
    raw_text = re.sub(r"URL Source:.*?\n", "", raw_text)  # Remove "URL Source: ..."
    raw_text = re.sub(r"Published Time:.*?\n", "", raw_text)  # Remove "Published Time: ..."
    raw_text = re.sub(r"Markdown Content:.*?\n", "", raw_text)  # Remove "Markdown Content: ..."
    
    # Step 2: Remove links, images, or anything resembling URL-like content
    raw_text = re.sub(r"https?://\S+", "", raw_text)  # Remove URLs
    raw_text = re.sub(r"!\[.*?\]\(.*?\)", "", raw_text)  # Remove images (like ![image](url))
    raw_text = re.sub(r"\[.*?\]\(.*?\)", "", raw_text)  # Remove markdown-style links (e.g., [link](URL))
    
    # Step 3: Clean extra markdown formatting, such as headings and bullet points
    raw_text = re.sub(r"#{1,6}\s*", "", raw_text)  # Remove markdown headings (##, ###, etc.)
    raw_text = re.sub(r"\*+\s*", "", raw_text)  # Remove bullet points (e.g., * item)
    raw_text = re.sub(r"\-+\s*", "", raw_text)  # Remove alternative bullet points (e.g., - item)
    raw_text = re.sub(r"\d+\.\s*", "", raw_text)  # Remove numbered lists (e.g., 1. item)
    
    # Step 4: Remove any residual elements such as text still inside square brackets or inconsistent characters
    raw_text = re.sub(r"\[.*?\]", "", raw_text)  # Remove any remaining square bracketed content
    raw_text = re.sub(r"={2,}", "", raw_text)  # Remove repeated equals signs used in markdown formatting (headers)
    
    # Step 5: Remove excessive newlines, empty paragraphs, and leading/trailing spaces
    raw_text = re.sub(r"\n{2,}", "\n", raw_text)  # Remove excessive newlines
    raw_text = re.sub(r"^\n+", "", raw_text)  # Remove leading newlines
    raw_text = re.sub(r"\n+$", "", raw_text)  # Remove trailing newlines
    
    # Step 6: Clean unwanted parentheses and any extraneous symbols
    raw_text = re.sub(r"[\(\)]", "", raw_text)  # Remove unwanted parentheses
    raw_text = re.sub(r"\s{2,}", " ", raw_text)  # Replace multiple spaces with single space
    raw_text = re.sub(r"\s*([.,;!?])\s*", r"\1", raw_text)  # Clean space around punctuation

    # Step 7: Final cleanup - remove extra spaces at the beginning and end
    raw_text = raw_text.strip()  # Strip extra spaces from the beginning and end

    return raw_text
```
